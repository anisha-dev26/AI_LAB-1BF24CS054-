```python
# File Name: USN_Week7_FOL_ForwardReasoning.py
# Description: Week 7 AI Lab - First Order Logic Forward Chaining Implementation

class KnowledgeBase:
    def __init__(self):
        # Initial facts represented as strings/tuples
        self.facts = {
            "American(Robert)",
            "Owns(A, T1)",
            "Missile(T1)",
            "Enemy(A, America)"
        }

    def forward_chain(self):
        print(r"Initial Knowledge Base Facts:")
        for f in self.facts:
            print(f" - {f}")
        print("\n--- Starting Forward Chaining Inference ---")

        new_facts_inferred = True
        step = 1

        while new_facts_inferred:
            new_facts_inferred = False
            inferred_this_iteration = []

            # Rule 1: Enemy(x, America) => Hostile(x)
            if "Enemy(A, America)" in self.facts and "Hostile(A)" not in self.facts:
                inferred_this_iteration.append("Hostile(A)")
                print(f"Step {step}: Inferred Hostile(A) from Enemy(A, America)")
                step += 1

            # Rule 2: Missile(x) => Weapon(x)
            if "Missile(T1)" in self.facts and "Weapon(T1)" not in self.facts:
                inferred_this_iteration.append("Weapon(T1)")
                print(f"Step {step}: Inferred Weapon(T1) from Missile(T1)")
                step += 1

            # Rule 3: Missile(x) ^ Owns(A, x) => Sells(Robert, x, A)
            if "Missile(T1)" in self.facts and "Owns(A, T1)" in self.facts and "Sells(Robert, T1, A)" not in self.facts:
                inferred_this_iteration.append("Sells(Robert, T1, A)")
                print(f"Step {step}: Inferred Sells(Robert, T1, A) from Missile(T1) and Owns(A, T1)")
                step += 1

            # Rule 4: American(p) ^ Weapon(q) ^ Sells(p,q,r) ^ Hostile(r) => Criminal(p)
            if {"American(Robert)", "Weapon(T1)", "Sells(Robert, T1, A)", "Hostile(A)"}.issubset(self.facts):
                if "Criminal(Robert)" not in self.facts:
                    inferred_this_iteration.append("Criminal(Robert)")
                    print(f"Step {step}: Inferred Criminal(Robert) from American, Weapon, Sells, and Hostile facts.")
                    step += 1

            if inferred_this_iteration:
                for fact in inferred_this_iteration:
                    self.facts.add(fact)
                new_facts_inferred = True

            # If goal reached, break early
            if "Criminal(Robert)" in self.facts:
                print("\n[SUCCESS] Goal 'Criminal(Robert)' has been proven!")
                break

        return "Criminal(Robert)" in self.facts

if __name__ == "__main__":
    kb = KnowledgeBase()
    kb.forward_chain()
```
