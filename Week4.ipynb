```python
import random

def h(b):
    n = len(b)
    return sum(b[i] == b[j] or abs(b[i]-b[j]) == j-i
               for i in range(n) for j in range(i+1, n))

def hill_climbing(n):
    b = [random.randrange(n) for _ in range(n)]
    step = 0

    while True:
        current = h(b)
        print("Step", step, ":", b, "Heuristic =", current)

        if current == 0:
            print("Solution found!")
            return b

        best = b[:]
        best_h = current

        for i in range(n):
            for r in range(n):
                t = b[:]
                t[i] = r
                if h(t) < best_h:
                    best, best_h = t, h(t)

        if best_h == current:
            print("Restarting...\n")
            return hill_climbing(n)

        b = best
        step += 1

n = int(input("Enter N: "))
if n < 1 or n in (2, 3):
    print("No solution exists")
else:
    b = hill_climbing(n)
    print("\nFinal Board:")
    for r in range(n):
        print(" ".join("Q" if b[c] == r else "." for c in range(n)))
```
